# Plotting the district Chikkamagaluru on the top of the DEM CE18

```python
import rasterio
import rasterio.plot
import geopandas as gpd
import numpy as np
import matplotlib.pyplot as plt
```

```python
src=rasterio.open(r"C:\Users\ASUS\Downloads\chikmagluruN13E075.SRTMGL1.hgt\N13E075.hgt")
print(src)
print("CRS", src.crs)
print ("Resolution:",src.res)
```

# Plotting Raster Data

```python
rasterio.plot.show(src)
```

# Customizing Plots

```python
fig,ax=plt.subplots(figsize=(8,8))
rasterio.plot.show(src, cmap="terrain", ax=ax, title="Digital Elevation Model (DEM)")
plt.show
```

```python
gdf=gpd.read_file(r"C:\Users\ASUS\Downloads\India_districts_shapefile.shp")
gdf_Chikkamaguluru=gdf[gdf['dtname']=='Chikkamagaluru']
gdf_Chikkamaguluru
```

```python
gdf_Chikkamaguluru=gdf_Chikkamaguluru.set_crs(src.crs)
```

```python
gdf_Chikkamaguluru.explore()
```

```python
fig,ax=plt.subplots(figsize=(8,8))
rasterio.plot.show(src, cmap="terrain", ax=ax, title="Digital Elevation Model (DEM)")
gdf_Chikkamaguluru.plot(ax=ax, edgecolor="purple", facecolor="none", linewidth=2)
```

# With Facecolor and alpha

```python
fig,ax=plt.subplots(figsize=(8,8))
rasterio.plot.show(src, cmap="terrain", ax=ax, title="Digital Elevation Model (DEM)")
gdf_Chikkamaguluru.plot(ax=ax, edgecolor="purple", facecolor="red", alpha=0.5, linewidth=2)
```
